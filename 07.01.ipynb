### 1. Рациональные числа

**Множество** - совокупность объектов, объединённых каким-либо общим признаком.

Числовые множества:

- **N** (*лат.*: naturalis - естественные, природный) - *множество натуральных чисел*: 1, 2, 3, ... Например, $2 \in \mathbb{N}$
- **Z** (*нем.*: Zahl - число) - *множество целых чисел*: ..., -1, 0, 1, 2, ... Например, Например, $-2 \in \mathbb{Z}$
- **Q** (*фр.*: quotient - отношение) - *множество рациональных чисел*: $\frac{1}{2}, \frac{3}{4}, -\frac{5}{6}$. Например, $ \frac{3}{4} \in \mathbb{Z}$
- **I* (*лат*.: irrationalis - неразумный) - *множество иррациональных чисел*: $\pi \approx 3,141592\dots$, $e \approx 2{,}71828\dots$
- **R** (лат.: realis - вещественные, действительные)

$-2 \notin \mathbb{N}$ - число -2 не принадлежит множеству **N**

$ \mathbb{N} ⊂ \mathbb{Z} ⊂ \mathbb{Q }⊂ \mathbb{R}$ - множества натуральных чисел, является подмножеством целых, являющихся подмножеством рациональных. Все они входя во множество действительных чисел.

> Любую обыкновенную дробь, можно представить представить в виде десятичной дроби.

Т.к. это рациональная дробь, то она либо конечная либо переодическая.

Обращение в конечную десятичную дробь:

$$
\begin{array}{r@{\;}r|l}
  & 1 & 8 \\
- & 10 & \overline{0{,}125} \\
  & \underline{8} & \\
- & 20 & \\
  & \underline{16} & \\
- & 40 & \\
  & \underline{40} & \\
  & 0 &
\end{array}
$$

Обращение в бесконечную десятичную дробь:

$$
\begin{array}{r|l}
  \underline{8} & 37 \\
- 80 & \overline{0{,}216216...} \\
  \underline{74} & \\
  60 & \\
- \underline{37} & \\
  230 & \\
- \underline{222} & \\
  80 & \\
- \underline{74} & \\
  60 & \\
- \underline{37} & \\
  230 & \\
- \underline{222} & \\
  8 &
\end{array}
$$

Т.к. при делении цифры начинают повторятся 216, то данные цифры называют периодом, а дробь переодической. Период такой дроби принято записывать в скобках: 0,(216)

Без полного вычисления десятичной можно поянять, будет ли она ялвяться переодичской или конечной:

- Если знаменатель есть только множители 2 и/или 5, то дробь будет конечной
- Если другие простые делители, то результат периодичная дробь

Каждое рациональное число можно представить в виде бесконечной десятичной дроби. Например дописав нули справа: 3,000...
