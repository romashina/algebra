12. Среди чисел 1458; 1805; 2342; 3620; 89217; 364425 найдите и выпишите те, которые: 

- а) делятся на 2;
- б) кратны 9;
- в) делятся на 5, но не кратны 3

**Ответы**:

- а) 1458, 2342, 3620;
- б) 1458, 89217;
- в) 1805, 3620

```python
twelve = [1458, 1805, 2342, 3620, 89217, 364425]
print(f"а)", *[i for i in twelve if i%2 == 0])
print(f"б)", *[i for i in twelve if i%9 == 0])
print(f"в)", *[i for i in twelve if i%5 == 0 and i%3 != 0])
```

13. Разложите на простые множители:

- а) 66;
- б) 1200;
- в) 5460;
- г) 1001

**Ответы**:

- а) 2 * 3 * 11
- б) 2 * 2 * 2 * 2 * 3 * 5 * 5
- в) 2 * 2 * 3 * 5 * 7 * 13
- г) 7 * 11 * 13

```python
from sympy import factorint

print(f"а)", *factorint(66).items())
print(f"б)", factorint(1200))
print(f"в)", factorint(5460))
print(f"г)", factorint(1001))
```
