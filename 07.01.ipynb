66. Сравните значения выражений:

- а) $2.06 \cdot 3.05$ и $21.28 \div 3.5$
- б) $97.2 \div 2.4$ и $62 - 21.6$
- в) $\frac{1}{2} + \frac{1}{5}$ и $\frac{1}{3} + \frac{1}{4}$
- г) $16 - 3\frac{5}{8}$ и $15 - 2\frac{1}{4}$


```python
def compare(x, y):
    # Приводим к float только для красивого вывода, если это Fraction
    fx = float(x)
    fy = float(y)
    
    # Форматируем с округлением до 2 знаков после запятой, отсекая лишние нули
    sx = f"{fx:.2f}".rstrip('0').rstrip('.')
    sy = f"{fy:.2f}".rstrip('0').rstrip('.')
    
    if x > y:
        return f"{sx} > {sy}"
    elif x < y:
        return f"{sx} < {sy}"
    else:
        return f"{sx} = {sy}"

    
print("Ответы:")
print("а)", compare(2.06 * 3.05, 21.28 / 3.5))
print("б)", compare(97.2 / 2.4, 62 - 21.6))
print("в)", compare(Fraction(1, 2) + Fraction(1, 5), Fraction(1, 3) + Fraction(1, 4)))
print("г)", compare(16 - (3 + Fraction(5, 8)), 15 - (2 + Fraction(1, 4))))
```

69. Сравните значения выражений:

- а) $0,7\cdot0,8\cdot0,9$ и $0,7 + 0,8 – 0,9$
- б) $\frac{1}{2}+\frac{1}{3}-\frac{1}{6}$ и $\frac{1}{2}\cdot\frac{1}{3}\cdot\frac{1}{6}$

```python
def compare(x, y):
    if x > y:
        return f"{x} > {y}"
    elif x < y:
        return f"{x} < {y}"
    else:
        return f"{x} = {y}"

print("а)", compare(round(0.7 * 0.8 * 0.9, 3), round(0.7 + 0.8 - 0.9, 3)))
print("б)", compare(Fraction(1, 2) + Fraction(1, 3) - Fraction(1, 6), 
                   Fraction(1, 2) * Fraction(1, 3) * Fraction(1, 6)))
```
